```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# --- Load Data for 60C ---
print("Please upload fctlab2.xlsx...")
files.upload() # Prompt for fctlab2.xlsx
df_60C = pd.read_excel('fctlab2.xlsx')
current_60C = df_60C['Current_Density (A/cm2)']
voltage_60C = df_60C['Voltage (V) at 60C']

# --- Load Data for 70C ---
print("Please upload pemfcell2.xlsx...")
files.upload() # Prompt for pemfcell2.xlsx
df_70C = pd.read_excel('pemfcell2.xlsx')
current_70C = df_70C['Current_Density (A/cm2)']
voltage_70C = df_70C['Voltage (V) at 70C']

# --- Load Data for 80C ---
print("Please upload pemfcell3.xlsx...")
files.upload() # Prompt for pemfcell3.xlsx
df_80C = pd.read_excel('pemfcell3.xlsx')
current_80C = df_80C['Current_Density (A/cm2)']
voltage_80C = df_80C['Voltage (V) at 80C']

# --- Create the combined plot ---
plt.figure(figsize=(10, 6))
plt.plot(current_60C, voltage_60C, label='Voltage at 60C', color='blue')
plt.plot(current_70C, voltage_70C, label='Voltage at 70C', color='red')
plt.plot(current_80C, voltage_80C, label='Voltage at 80C', color='green')

plt.xlabel('Current Density (A/cm2)')
plt.ylabel('Voltage (V)')
plt.title('Current vs Voltage at Different Temperatures')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Calculate power for each temperature
power_60C = current_60C * voltage_60C
power_70C = current_70C * voltage_70C
power_80C = current_80C * voltage_80C

# Create the combined power plot
plt.figure(figsize=(10, 6))
plt.plot(current_60C, power_60C, label='Power at 60C', color='blue')
plt.plot(current_70C, power_70C, label='Power at 70C', color='red')
plt.plot(current_80C, power_80C, label='Power at 80C', color='green')

plt.xlabel('Current Density (A/cm2)')
plt.ylabel('Power (W)')
plt.title('Power vs Current at Different Temperatures')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot Voltage on primary y-axis (ax1)
ax1.plot(current_60C, voltage_60C, label='Voltage at 60C', color='blue', linestyle='-')
ax1.plot(current_70C, voltage_70C, label='Voltage at 70C', color='red', linestyle='-')
ax1.plot(current_80C, voltage_80C, label='Voltage at 80C', color='green', linestyle='-')
ax1.set_xlabel('Current Density (A/cm2)')
ax1.set_ylabel('Voltage (V)', color='black')
ax1.tick_params(axis='y', labelcolor='black')

# Create a secondary y-axis for Power
ax2 = ax1.twinx()
ax2.plot(current_60C, power_60C, label='Power at 60C', color='blue', linestyle='-')
ax2.plot(current_70C, power_70C, label='Power at 70C', color='red', linestyle='-')
ax2.plot(current_80C, power_80C, label='Power at 80C', color='green', linestyle='-')
ax2.set_ylabel('Power (W)', color='darkgray')
ax2.tick_params(axis='y', labelcolor='darkgray')

# Combine legends from both axes
handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(handles1 + handles2, labels1 + labels2, loc='best')

plt.title('Voltage and Power vs Current Density at Different Temperatures')
plt.grid(True)
# plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax1 = plt.subplots(figsize=(10, 6))

# Plot Voltage at 80C on primary y-axis (ax1)
ax1.plot(current_80C, voltage_80C, label='Voltage at 80C', color='green', linestyle='-')
ax1.set_xlabel('Current Density (A/cm2)')
ax1.set_ylabel('Voltage (V)', color='black')
ax1.tick_params(axis='y', labelcolor='black')

# Create a secondary y-axis for Power at 80C
ax2 = ax1.twinx()
ax2.plot(current_80C, power_80C, label='Power at 80C', color='blue', linestyle='-')
ax2.set_ylabel('Power (W)', color='darkgray')
ax2.tick_params(axis='y', labelcolor='darkgray')

# Combine legends from both axes
handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(handles1 + handles2, labels1 + labels2, loc='best')

plt.title('Voltage and Power vs Current Density at 80C')
plt.grid(True)
plt.tight_layout()
plt.show()
```
